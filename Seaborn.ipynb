```python
import seaborn as sns
import matplotlib.pyplot as plt
```

### distplot()  histplot() both are same   ) bins by default 20 and we can set any value 
### Kernal density estimeter kde=True then we plot a curve line es sy data ki skew ness pta chlty ha ky data hmra kis side sy zyda skew ya normal distributed ha 

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],bins=20 , kde=True)
plt.subplot(1,2,2)
sns.histplot(df['tip'],bins=20 , kde=True)
```

### jointplot() kind of plot thes are 1) hex 2) regression 

```python
sns.jointplot(x='total_bill', y='tip', data=df, kind='scatter')
```

```python
sns.jointplot(x='total_bill', y='tip', data=df, kind='reg')
```

```python
sns.pairplot(df)
```

```python
sns.pairplot(df, hue='sex')
```

```python
sns.rugplot(df)
```
